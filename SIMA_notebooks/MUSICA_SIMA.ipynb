```python
#Import statements
import numpy as np
import uxarray as ux
import cartopy.crs as ccrs
import holoviews as hv
```

```python
#Needed mesh file(s):

mesh_file = '/glade/campaign/cesm/cesmdata/inputdata/share/meshes/ne3pg3_ESMFmesh_c221214_cdf5.nc'

#Needed SIMA history file(s):

cam_sima_data_file = 'scripts/chemistry-demo.cam.h1i.0001-01-01-00000.nc'
```

```python
#Open dataset:
uxds = ux.open_dataset(mesh_file, cam_sima_data_file)
```

```python
#Extract relevant variables
Cl = uxds['Cl']
Cl2 = uxds['Cl']
Foo = uxds['Foo']
```

```python
Cl2
```

```python
Cl[0,19,:].plot(
    height=400,
    width=900,
    clabel="",
    backend="matplotlib",
    cmap="Blues",
    projection=ccrs.PlateCarree(central_longitude=180),
    features=["borders", "coastline"],
    title="Cl at ~500 hPa"
)
```

```python
#Plot Cl2
Cl2[0,19,:].plot(
    height=400,
    width=900,
    clabel="",
    backend="matplotlib",
    cmap="plasma",
    projection=ccrs.PlateCarree(central_longitude=180),
    #projection=ccrs.Robinson(central_longitude=180),
    features=["borders", "coastline"],
    title="Cl2 at ~500 hPa"
)
```
